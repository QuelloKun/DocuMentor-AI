```python
# Test basic arXiv search
client = arxiv.Client()

# Search for a few recent ML papers
search = arxiv.Search(
    query="cat:cs.LG",
    max_results=5,
    sort_by=arxiv.SortCriterion.SubmittedDate
)

print("🔍 Testing arXiv API with recent Machine Learning papers:")
print("=" * 60)

for i, result in enumerate(client.results(search), 1):
    print(f"{i}. {result.title}")
    print(f"   Authors: {', '.join([author.name for author in result.authors[:3]])}{'...' if len(result.authors) > 3 else ''}")
    print(f"   Published: {result.published.strftime('%Y-%m-%d')}")
    print(f"   Categories: {', '.join(result.categories)}")
    print(f"   PDF URL: {result.pdf_url}")
    print()

print("✅ arXiv API connection successful!")
```
